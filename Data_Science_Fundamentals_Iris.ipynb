## 1. Understand the Dataset
The Iris dataset has 150 flowers from 3 species (setosa, versicolor, virginica) with 4 measurements each (cm). Business/scientific context: classify a species from measurable features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

sns.set_style('whitegrid')
pd.set_option('display.max_columns', None)

iris = load_iris(as_frame=True)
df = iris.frame.copy()
df['species'] = df['target'].map(dict(enumerate(iris.target_names)))
df = df.drop(columns='target')

print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
# Value counts per species (class balance)
df['species'].value_counts()
```

## 2. Apply Python Fundamentals
Data structures, functions, loops, conditionals, and libraries (Pandas, NumPy).

```python
# Data structure: dictionary of feature -> (min, max)
features = [c for c in df.columns if c != 'species']
ranges = {col: (df[col].min(), df[col].max()) for col in features}
for col, (lo, hi) in ranges.items():
    print(f"{col}: {lo} to {hi}")
```

```python
# Function + conditional: categorize a flower by petal length
def petal_size(length):
    if length < 2:
        return 'Small'
    elif length < 5:
        return 'Medium'
    else:
        return 'Large'

df['petal_size'] = df['petal length (cm)'].apply(petal_size)
df['petal_size'].value_counts()
```

```python
# Loop + NumPy: mean and standard deviation of each feature
for col in features:
    arr = df[col].to_numpy()
    print(f"{col}: mean={np.mean(arr):.3f}, std={np.std(arr, ddof=1):.3f}")
```

## 3. Data Cleaning
Check missing values, duplicates, inconsistent data, and data types.

```python
print("Missing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
print("\nSpecies labels:", df['species'].unique())
```

```python
# Remove duplicate records
before = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print(f"Removed {before - len(df)} duplicate row(s). New shape: {df.shape}")
```

## 4. Statistical Concepts
Mean, median, mode, standard deviation, variance, and correlation.

```python
stats = pd.DataFrame({
    'mean': df[features].mean(),
    'median': df[features].median(),
    'mode': df[features].mode().iloc[0],
    'std': df[features].std(),
    'variance': df[features].var()
}).round(3)
stats
```

```python
# Correlation matrix
corr = df[features].corr().round(3)
corr
```

```python
# Mean of each feature by species
df.groupby('species')[features].mean().round(3)
```

**Interpretation:** petal length and petal width are very strongly correlated (r ≈ 0.96) and have the largest spread, so they vary most across species. Sepal width is weakly (negatively) correlated with the other features and varies least.

## 5. Exploratory Data Analysis & Visualizations

```python
# Histograms of each feature
df[features].hist(bins=15, figsize=(10,7), color='#4C72B0', edgecolor='white')
plt.suptitle('Distribution of Iris Features', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Box plots for outlier detection
fig, axes = plt.subplots(1, 4, figsize=(14,5))
for ax, col in zip(axes, features):
    sns.boxplot(y=df[col], ax=ax, color='#8CA0D7')
    ax.set_title(col, fontsize=9)
plt.suptitle('Boxplots - Outlier Detection', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(6.5,5.5))
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
plt.title('Correlation Heatmap of Iris Features')
plt.show()
```

```python
# Scatter plot: petal length vs petal width by species
plt.figure(figsize=(7,5.5))
sns.scatterplot(data=df, x='petal length (cm)', y='petal width (cm)', hue='species', s=50, edgecolor='k')
plt.title('Petal Length vs Petal Width by Species')
plt.show()
```

```python
# Bar chart: mean feature values by species
df.groupby('species')[features].mean().plot(kind='bar', figsize=(8,5.5))
plt.ylabel('Mean value (cm)')
plt.title('Mean Feature Values by Species')
plt.xticks(rotation=0)
plt.show()
```

## 6. Analysis Report — Key Insights & Conclusions

- The dataset is clean and balanced: no missing values, 3 species with 50 samples each; one duplicate row was found and removed.
- Petal length and petal width are the most informative features: highest variance, strongly correlated (r ≈ 0.96), and they separate the three species clearly.
- Setosa forms a tight, completely separate cluster; versicolor and virginica overlap partially, so a multi-feature rule works better than a single threshold.
- Sepal width behaves differently from the other features (weak negative correlations, lowest variance) and is the weakest single discriminator.
- **Recommendation:** prioritize petal length and petal width as primary features for any classification rule or model, with sepal measurements as supporting features.
